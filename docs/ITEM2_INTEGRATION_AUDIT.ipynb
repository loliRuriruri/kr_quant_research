## Context & Methods
동일 세대와 신호 ID로 비교합니다. 기본 캘린더 정렬은 pre_entry_rank이며 전송 배열 순서가 아닙니다. 원천 입력은 현재 로컬 저장본입니다. 과거 수익성/OOS 검증은 아닙니다.

```python
from pathlib import Path
import runpy
root = Path.cwd()
while not (root / 'scripts/audit_item2_integration.py').exists():
    if root == root.parent:
        raise RuntimeError('저장소 또는 docs 디렉터리에서 실행하세요.')
    root = root.parent
audit = runpy.run_path(str(root / 'scripts/audit_item2_integration.py'))
result = audit['run'](probe_source=False)
```

```python
assert all(value is True for value in result['checks'].values()), result['checks']
print('연동 검사:', result['checks'])
print('시점:', result['freshness'])
print('배포 준비:', result['publication'])
```

## Takeaways
기준일 차이가 있으면 데이터를 최신이라고 간주하지 않습니다. 공식 원천 확인은 스크립트 --probe-source 옵션으로 별도 실행할 수 있습니다. 이 노트북에는 네트워크 조회가 없는 기본 경로만 포함했습니다. 메뉴의 외부 자료 요청 및 자동 수집은 시작하지 않습니다. 정식 Jupyter 커널 의존성은 현재 미설치이며 코드 셀 순서 실행으로만 검사했습니다.
